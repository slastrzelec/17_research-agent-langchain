```python
from dotenv import load_dotenv
import os

load_dotenv()

# sprawdzenie czy klucz działa
print("key loaded:", bool(os.getenv("OPENAI_API_KEY")))
```

```python
from langchain_community.tools import WikipediaQueryRun
from langchain_community.utilities import WikipediaAPIWrapper

wiki_tool = WikipediaQueryRun(api_wrapper=WikipediaAPIWrapper())

# test
result = wiki_tool.run("carbon nanotubes")
print(result)
```

```python
from langchain_openai import ChatOpenAI
from langgraph.prebuilt import create_react_agent

llm = ChatOpenAI(model="gpt-4o-mini", temperature=0)
tools = [wiki_tool]

agent_executor = create_react_agent(llm, tools)
```

```python

result = agent_executor.invoke({
    "messages": [("user", "What are the medical applications of carbon nanotubes?")]
})

print(result["messages"][-1].content)
```

```python
for step in agent_executor.stream({
    "messages": [("user", "What are the medical applications of carbon nanotubes?")]
}):
    print(step)
    print("---")
```

```python
import sys
sys.path.append("..")  # use the same safe calculator as the app
from langchain_core.tools import tool
from safe_calc import safe_eval, CalcError

@tool
def calculate(expression: str) -> str:
    """Useful for mathematical calculations. Input should be a mathematical expression."""
    try:
        return str(safe_eval(expression))
    except CalcError as e:
        return f"Error: {e}"

tools = [wiki_tool, calculate]
agent_executor = create_react_agent(llm, tools)
```

```python
result = agent_executor.invoke({
    "messages": [("user", "What is the diameter of SWCNT in meters? Convert 1.5 nanometers to meters.")]
})
print(result["messages"][-1].content)
```
